# AC820 RiceVision — 一键启动

运行下面唯一的启动单元，将初始化摄像头、LCD、触摸、模型与网页预览。
重复运行会返回已有实例。需要更改参数时先停止应用。


```python
from pathlib import Path
import sys
for candidate in (Path.cwd(), *Path.cwd().parents):
    if (candidate/'lanuch.py').is_file() and (candidate/'src/trash_pynq').is_dir():
        ROOT=candidate
        break
else:
    raise RuntimeError('Open this notebook inside the pynq_port folder')
if str(ROOT) not in sys.path:
    sys.path.insert(0,str(ROOT))
import lanuch
app=lanuch.start()
```
